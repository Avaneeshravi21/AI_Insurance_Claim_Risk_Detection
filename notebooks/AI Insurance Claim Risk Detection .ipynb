```python
Detect outliers using IQR
```

```python
outlier_summary = []

for col in numeric_features:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    
    IQR = Q3 - Q1
    
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    
    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ][col].count()
    
    outlier_summary.append({
        "Feature": col,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outliers
    })

outlier_summary = pd.DataFrame(outlier_summary)

outlier_summary
```

```python
from sklearn.model_selection import StratifiedKFold, cross_validate
```

```python
  # Create Stratified K-Fold cross-validation

skf = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)  
```

```python
# Define cross-validation scoring metrics

cv_scoring = {
    'accuracy': 'accuracy',
    'precision': 'precision',
    'recall': 'recall',
    'f1': 'f1',
    'roc_auc': 'roc_auc'
}
```

```python
from sklearn.pipeline import Pipeline

logistic_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        (
            'model',
            LogisticRegression(
                class_weight=balanced_weights,
                max_iter=1000,
                random_state=42
            )
        )
    ]
)
```

```python
random_forest_pipeline = Pipeline(
    steps=[
        ('preprocessor', preprocessor),
        (
            'model',
            RandomForestClassifier(
                n_estimators=200,
                class_weight=balanced_weights,
                random_state=42,
                n_jobs=-1
            )
        )
    ]
)
```

```python
# Cross-validation for Logistic Regression

logistic_cv = cross_validate(
    logistic_pipeline,
    X_structured,
    y,
    cv=skf,
    scoring=cv_scoring,
    n_jobs=-1
)
```

```python
# Display Logistic Regression CV results

logistic_cv_results = pd.DataFrame({
    'Fold': range(1, 6),
    'Accuracy': logistic_cv['test_accuracy'],
    'Precision': logistic_cv['test_precision'],
    'Recall': logistic_cv['test_recall'],
    'F1 Score': logistic_cv['test_f1'],
    'ROC-AUC': logistic_cv['test_roc_auc']
})

logistic_cv_results.round(4)
```

```python
# Calculate average Logistic Regression CV performance

logistic_cv_mean = {
    'Accuracy': logistic_cv['test_accuracy'].mean(),
    'Precision': logistic_cv['test_precision'].mean(),
    'Recall': logistic_cv['test_recall'].mean(),
    'F1 Score': logistic_cv['test_f1'].mean(),
    'ROC-AUC': logistic_cv['test_roc_auc'].mean()
}

pd.Series(logistic_cv_mean).round(4)
```

```python
# Cross-validation for Random Forest

random_forest_cv = cross_validate(
    random_forest_pipeline,
    X_structured,
    y,
    cv=skf,
    scoring=cv_scoring,
    n_jobs=-1
)
```

```python
# Cross-validation for Random Forest

random_forest_cv = cross_validate(
    random_forest_pipeline,
    X_structured,
    y,
    cv=skf,
    scoring=cv_scoring,
    n_jobs=-1
)
```

```python
# Display Random Forest CV results

random_forest_cv_results = pd.DataFrame({
    'Fold': range(1, 6),
    'Accuracy': random_forest_cv['test_accuracy'],
    'Precision': random_forest_cv['test_precision'],
    'Recall': random_forest_cv['test_recall'],
    'F1 Score': random_forest_cv['test_f1'],
    'ROC-AUC': random_forest_cv['test_roc_auc']
})

random_forest_cv_results.round(4)
```

```python
def predict_claim_risk(claim_data):
    
    # Transform raw claim data
    claim_processed = preprocessor.transform(claim_data)
    
    # Generate probability
    probability = random_forest_model.predict_proba(
        claim_processed
    )[:, 1][0]
    
    # Generate prediction
    prediction = random_forest_model.predict(
        claim_processed
    )[0]
    
    # Generate risk category
    risk_category = get_risk_category(probability)
    
    return {
        "prediction": int(prediction),
        "probability": float(probability),
        "risk_category": risk_category
    }
```
